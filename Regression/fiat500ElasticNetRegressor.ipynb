```python
dff=df.drop(['ID','lat','lon'],axis=1)
```

```python
dff=pd.get_dummies(dff,dtype=int)
```

```python
x=dff.drop(['price'],axis=1)
y=dff['price']
```

```python
from sklearn.metrics import r2_score,mean_squared_error
from sklearn.model_selection import train_test_split
```

```python
xtrain,xtest,ytrain,ytest=train_test_split(x,y,test_size=0.20,random_state=42)
```

```python
# elasticnet
```

```python
from sklearn.linear_model import ElasticNet
from sklearn.model_selection import GridSearchCV
elastic=ElasticNet()
parameters={'alpha':[1e-15,1e-10,1e-8,1e-4,1e-3,1e-2,1,5,10,20]}
elastic_reg=GridSearchCV(elastic,parameters)
elastic_reg.fit(xtrain,ytrain)
```

```python
elastic_reg.best_params_
```

```python
elastic=ElasticNet(alpha=0.01)
elastic.fit(xtrain,ytrain)
ypred_elastic=elastic.predict(xtest)
```

```python
r2=r2_score(ytest,ypred_elastic)
r2
```

```python
mse=mean_squared_error(ytest,ypred_elastic)
mse
```
