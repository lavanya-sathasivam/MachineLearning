```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df=pd.read_csv(r"C:\Users\lavan\Desktop\MachineLearning\uber.csv")
```

```python
df['pickup_datetime']=pd.to_datetime(df['pickup_datetime'])
```

```python
df['Year']=df['pickup_datetime'].dt.year
df['Month']=df['pickup_datetime'].dt.month
df['Date']=df['pickup_datetime'].dt.day
df['Hour']=df['pickup_datetime'].dt.hour
```

```python
df.head()
```

```python
#yearwise revenue sum and plot

year_revenue=df.groupby('Year')['fare_amount'].sum()
print(year_revenue)
year_revenue.plot(kind='bar')
plt.show()
```

```python
#monthwise revenue sum and plot

month_revenue=df.groupby('Month')['fare_amount'].sum()
print(month_revenue)
month_revenue.plot(kind='bar')
plt.show()
```

```python
def get_date(day):
    if day>0 and day<=10:
        return '1-10'
    elif day>10 and day<=20:
        return '11-20'
    else:
        return '21-31'
df['date_slot']=df['Date'].apply(get_date)
```

```python
c=df.groupby('date_slot').size()
print(c)
c.plot(kind='bar')
plt.show()
```

```python
def get_slot(hour):
    if 0<=hour and hour<6:
        return '0-6'
    elif 6<=hour and hour< 12:
        return '6-12'
    elif 12<=hour and hour<18:
        return '12-18'
    else:
        return '18-24'
df['Time_Slot']=df['Hour'].apply(get_slot)
```

```python
service_count=df.groupby('Time_Slot').size()
print(service_count)
service_count.plot(kind='bar')
plt.show()
```

```python
passenger_year_revenue = df.pivot_table(
    values='fare_amount',
    index='passenger_count',
    columns='Year',
    aggfunc='sum'
)

print(passenger_year_revenue)

passenger_year_revenue.plot(kind='bar', figsize=(10,6))
plt.title('Passenger Count Wise Revenue by Year')
plt.ylabel('Revenue')
plt.show()
```

```python
passenger_month_revenue = df.pivot_table(
    values='fare_amount',
    index='passenger_count',
    columns='Month',
    aggfunc='sum'
)

print(passenger_month_revenue)

passenger_month_revenue.plot(kind='bar', figsize=(12,6))
plt.title('Passenger Count Wise Revenue by Month')
plt.ylabel('Revenue')
plt.show()
```

```python
result = []

for year in sorted(df['Year'].unique()):
    for month in range(1, 13):
        revenue = df[(df['Year']==year)&(df['Month'] == month)]['fare_amount'].sum()
        result.append([year, month, revenue])
revenue_df = pd.DataFrame(
    result,
    columns=['Year', 'Month', 'Revenue']
)
revenue_df.to_csv(r"C:\Users\lavan\Desktop\MachineLearning\year.csv")
```
